## 1. 定義修正後的批次激發函數與損失函數

```python
import numpy as np


def relu(x):
    return np.maximum(0, x)


def sigmoid(x):
    return 1 / (1 + np.exp(-x))


def softmax(x):
    if x.ndim == 2:
        x = x.T
        x = x - np.max(x, axis=0)
        y = np.exp(x) / np.sum(np.exp(x), axis=0)
        return y.T
    return np.exp(x) / np.sum(np.exp(x))


def cross_entropy_error(y_hat, y):
    if y_hat.ndim == 1:
        y = y.reshape(1, -1)
        y_hat = y_hat.reshape(1, -1)
    batch_size = y_hat.shape[0]
    delta = 1e-7
    return -np.sum(y * np.log(y_hat + delta)) / batch_size
```

## 2. 批次資料產生與網路參數初始化

```python

np.random.seed(42)
X = np.random.randn(5, 6)
Y = np.array([
    [1, 0, 0],
    [0, 1, 0],
    [0, 0, 1],
    [1, 0, 0],
    [0, 1, 0]
])

W1 = np.random.randn(6, 5) * 0.1
b1 = np.random.randn(5) * 0.1

W2 = np.random.randn(5, 4) * 0.1
b2 = np.random.randn(4) * 0.1

W3 = np.random.randn(4, 4) * 0.1
b3 = np.random.randn(4) * 0.1

W4 = np.random.randn(4, 3) * 0.1
b4 = np.random.randn(3) * 0.1

print("【資料與參數陣列形狀列出】")
print(f"輸入矩陣 X 形狀: {X.shape}")
print(f"標籤矩陣 Y 形狀: {Y.shape}")
print(f"W1 形狀: {W1.shape} | b1 形狀: {b1.shape}")
print(f"W2 形狀: {W2.shape} | b2 形狀: {b2.shape}")
print(f"W3 形狀: {W3.shape} | b3 形狀: {b3.shape}")
print(f"W4 形狀: {W4.shape} | b4 形狀: {b4.shape}")
```

## 3. 批次前向傳播計算與陣列形狀顯示

```python

Z1 = np.dot(X, W1) + b1
H1 = relu(Z1)

Z2 = np.dot(H1, W2) + b2
H2 = relu(Z2)

Z3 = np.dot(H2, W3) + b3
H3 = sigmoid(Z3)

Z4 = np.dot(H3, W4) + b4
Y_hat = softmax(Z4)

print("【前向傳播矩陣結果與形狀】\n")
print(f"Z1 形狀: {Z1.shape}\n數值:\n{Z1}\n")
print(f"H1 形狀: {H1.shape}\n數值:\n{H1}\n")
print("-"*60)
print(f"Z2 形狀: {Z2.shape}\n數值:\n{Z2}\n")
print(f"H2 形狀: {H2.shape}\n數值:\n{H2}\n")
print("-"*60)
print(f"Z3 形狀: {Z3.shape}\n數值:\n{Z3}\n")
print(f"H3 形狀: {H3.shape}\n數值:\n{H3}\n")
print("-"*60)
print(f"Z4 形狀: {Z4.shape}\n數值:\n{Z4}\n")
print(f"最終預測 Y_hat 形狀: {Y_hat.shape}\n數值:\n{Y_hat}\n")
```

## 4. 計算批次平均交叉熵誤差

```python
loss_average = cross_entropy_error(Y_hat, Y)
print("=== 批次誤差評估 ===")
print(f"五筆批次資料的平均交叉熵誤差 (Average Loss): {loss_average:.6f}")
```
